A better way is **imputation**, which means filling in the missing values with a calculated guess.Common imputation strategies:*   **Mean:** Fill with the average value. Good for normally distributed data.*   **Median:** Fill with the middle value. Better for skewed data or data with outliers (like `Fare`).*   **Mode:** Fill with the most frequent value. Used for categorical data.

## Part 4: Feature Scaling**Concept:** Many models are sensitive to the scale of the features. For example, `Age` (from 0-80) and `Fare` (from 0-512) are on very different scales. This can cause the model to incorrectly believe that `Fare` is a more important feature simply because its values are larger.**Feature Scaling** solves this by putting all features on a similar scale. A common method is **Standardization** (`StandardScaler` in scikit-learn), which rescales the data to have a mean of 0 and a standard deviation of 1.**Important:** You only scale your numerical features, not your target variable or your newly encoded categorical columns.

### Task 3: Scale the 'Age' and 'Fare' Columns**Your Task:** Use `StandardScaler` from `sklearn.preprocessing` to scale the 'Age' and 'Fare' columns.

1. Why is it often better to impute missing values with the median instead of the mean?
Imputing with the median is often preferred because the median is far more robust to outliers and skewed distributions. A few extremely large or small values can pull the mean in their direction, producing an imputation value that doesn’t represent the typical observation. The median, on the other hand, reflects the central tendency of the data without being distorted by extreme values, making it a more stable and reliable choice for many real‑world datasets.

2. Explain in your own words what One‑Hot Encoding does and why it is necessary.
One‑Hot Encoding transforms categorical variables into a set of binary indicator columns, where each new column represents one category and takes the value 1 if the observation belongs to that category and 0 otherwise. This transformation is necessary because most machine learning algorithms cannot directly interpret text labels; they require numerical inputs. One‑Hot Encoding preserves the categorical information without introducing false ordinal relationships, ensuring that the model treats each category as distinct rather than implying any ranking between them.

3. Would you need to apply Feature Scaling to a Decision Tree model? Why or why not?
Feature scaling is generally unnecessary for Decision Trees because they do not rely on distance‑based calculations or gradient‑based optimization. A tree makes splits by evaluating thresholds on individual features, so the scale of the feature does not influence how the splits are chosen. Whether a feature ranges from 0 to 1 or 0 to 10,000, the tree considers each feature independently and selects the split that maximizes information gain or reduces impurity. As a result, scaling has no meaningful effect on the structure or performance of a Decision Tree.
